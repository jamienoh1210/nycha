```python
import pandas as pd
import geopandas as gpd

import matplotlib.pyplot as plt
import seaborn as sns


pluto_raw_path = '../data/raw/Primary_Land_Use_Tax_Lot_Output_(PLUTO)_20260314.csv'

pluto = pd.read_csv(pluto_raw_path)

pluto.head()
```

```python
pluto_gdf = gpd.GeoDataFrame(
  pluto,
  geometry = gpd.points_from_xy(pluto['longitude'], pluto['latitude'], crs="EPSG:2263"))

pluto_gdf.head()

```

```python
cleaned_pluto_gdf = pluto_gdf.copy()

# Standardize column names
cleaned_pluto_gdf.columns= (cleaned_pluto_gdf.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

cleaned_pluto_gdf.head()
```

```python

keep = ['geometry','borough', 'tax_block', 'tax_lot', 'census_tract_2010', 'cb2010', 'council_district', 'postcode', 'policeprct', 'healtharea', 'address', 'zonedist1', 'zonedist2', 'zonedist3', 'zonedist4', 'ltdheight', 'bldgclass', 'landuse', 'easements', 'ownertype', 'ownername', 'lotarea', 'bldgarea', 'resarea', 'garagearea', 'strgearea', 'otherarea', 'areasource', 'numbldgs', 'numfloors', 'unitsres', 'unitstotal', 'lotfront', 'lotdepth', 'bldgfront', 'bldgdepth', 'proxcode', 'irrlotcode', 'lottype', 'assessland', 'assesstot', 'yearbuilt', 'yearalter1', 'yearalter2', 'histdist', 'landmark', 'builtfar', 'residfar', 'borocode', 'bbl', 'tract2010', 'xcoord', 'ycoord', 'latitude', 'longitude', 'zonemap', 'zmcode', 'appbbl', 'appdate', 'plutomapid', 'dcpedited', 'notes', 'bct2020', 'bctcb2020']


dropped_pluto_gdf = cleaned_pluto_gdf[keep]

dropped_pluto_gdf = dropped_pluto_gdf.copy()

dropped_pluto_gdf["yearbuilt_dt"] = pd.to_datetime(
    dropped_pluto_gdf["yearbuilt"],
    format="%Y",
    errors="coerce"
)

dropped_pluto_gdf["yearalter1_dt"] = pd.to_datetime(
    dropped_pluto_gdf["yearalter1"].str.replace(",", ""),
    format="%Y",
    errors="coerce"
)


dropped_pluto_gdf["yearalter2_dt"] = pd.to_datetime(
    dropped_pluto_gdf["yearalter2"].str.replace(",", ""),
    format="%Y",
    errors="coerce"
)




dropped_pluto_gdf.info()
#pluto_owner_types = dropped_pluto_gdf.groupby("ownertype").size()
#pluto_owner_types.head()

#note from talking to Johnny: check type for each column and make sure they make sense.
```

```python
pluto_owner_types = (
    dropped_pluto_gdf.groupby("ownername")
    .size()                     # count parcels per owner
    .sort_values(ascending=False)  # largest first
)

# Top 10 owners
pluto_owner_types.head(10)
```

```python
# List of owners to keep
owners_to_keep = [
    "NYC HOUSING AUTHORITY",
    "NYC HOUSING PRESERVATION AND DEVELOPMENT"
]

# Filter the GeoDataFrame
nycha_housing_gdf = dropped_pluto_gdf[
    dropped_pluto_gdf["ownername"].isin(owners_to_keep)
].copy()  # copy() avoids SettingWithCopyWarning

nycha_housing_gdf.head(20)

```

```python
nycha_housing_gdf.to_csv('../data/processed/cleaned_PLUTO.csv', index=False)
```
