```python
import pandas as pd
import geopandas as gpd

import matplotlib.pyplot as plt
import seaborn as sns

nycha_dev_data_raw_path = '../data/raw/NYCHA_Development_Data_Book_20260314.csv'
nycha_dev_data = pd.read_csv(nycha_dev_data_raw_path)

nycha_dev_data.info()
nycha_dev_data.head()
```

```python
num_unique_values = nycha_dev_data['DATA AS OF'].nunique()
print(f"Number of unique values in 'DATA AS OF' column: {num_unique_values}")
```

```python
nycha_dev_data_clean = nycha_dev_data.drop(columns=['OPERATING EDP#','DEVELOPMENT EDP#','HUD AMP#','AVG NO R/R PER APARTMENT','POPULATION SECTION 8 TRANSITION','TOTAL # OF FIXED INCOME HOUSEHOLD','NUMBER OF NON-RESIDENTIAL BLDGS','NUMBER OF STAIRHALLS','NET DEV AREA SQ FT','ACRES','EXCLUDING PARK ACRES','CUBAGE CU FT','BLDG COVERAGE %','PER RENTAL ROOM','LOCATION STREET A','LOCATION STREET B','LOCATION STREET C','LOCATION STREET D','COMMUNITY DISTIRCT','US CONGRESSIONAL DISTRICT','NY STATE SENATE DISTRICT','NY STATE ASSEMBLY DISTRICT','NY CITY COUNCIL DISTRICT'])
```

```python
nycha_dev_data_clean.columns = (nycha_dev_data_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace('#',""))


##nycha_dev_data_clean.head(15,)
nycha_dev_data_clean.info()
```

```python
import pandas as pd

nycha_dev_data_clean['federalized_development'] = pd.to_datetime(
    nycha_dev_data_clean['federalized_development'],
    errors='coerce'
)

nycha_dev_data_clean['completion_date'] = pd.to_datetime(
    nycha_dev_data_clean['completion_date',
    errors='coerce'
)


```

```python
nycha_dev_data_clean['rad_transferred_date'] = pd.to_datetime(
    nycha_dev_data_clean['rad_transferred_date'],
    errors='coerce'
)
```
