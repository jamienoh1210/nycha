```python
import pandas as pd
import geopandas as gpd

nychares_raw_path = '/content/drive/MyDrive/Informatics Final Project/Data_Sources/raw_data/NYCHA_Residential_Addresses_20260314.csv'

nychares = pd.read_csv(nychares_raw_path)

nychares.head()
```

```python
nychares_gdf = gpd.GeoDataFrame(
  nychares,
  geometry = gpd.points_from_xy(nychares['LONGITUDE'], nychares['LATITUDE'], crs="EPSG:2263"))

nychares_gdf.head()
```

```python
cleaned_nyres_gdf = nychares_gdf.copy()

# Standardize column names
cleaned_nyres_gdf.columns= (cleaned_nyres_gdf.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_"))


cleaned_nyres_gdf.head()
```

```python
cleaned_nyres_gdf.to_csv('/content/drive/MyDrive/Informatics Final Project/Data_Sources/raw_data/nycha_residential.csv', index = False)
```
